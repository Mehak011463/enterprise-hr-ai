# Feature engineering
The final model uses a sklearn preprocessing pipeline so raw employee features can be passed to inference without manually creating one-hot columns.

```python
import pandas as pd
df=pd.read_csv('../data/raw/employee_attrition.csv')
df['Attrition_Target']=df['Attrition'].map({'Yes':1,'No':0})
df['Income_Per_Company_Year']=df['MonthlyIncome']/(df['YearsAtCompany']+1)
df['Overall_Satisfaction']=(df['EnvironmentSatisfaction']+df['JobSatisfaction']+df['RelationshipSatisfaction'])/3
print(df[['Income_Per_Company_Year','Overall_Satisfaction','Attrition_Target']].head())
```
