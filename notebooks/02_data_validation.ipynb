```python
import pandas as pd
import numpy as np

DATA_PATH = "../data/raw"

# Load raw datasets
df_attrition = pd.read_csv(f"{DATA_PATH}/employee_attrition.csv")
df_perf = pd.read_csv(f"{DATA_PATH}/Employee_Performance_Dataset.csv")
df_occ = pd.read_csv(f"{DATA_PATH}/occupation_data.csv")
df_ess = pd.read_csv(f"{DATA_PATH}/essential_skills.csv")
df_soft = pd.read_csv(f"{DATA_PATH}/software_skills.csv")

print("--- Running Data Validation Checks ---")

# 1. Attrition Data Rules
assert df_attrition['Age'].between(18, 100).all(), "Validation Error: Age out of 18-100 bound"
assert df_attrition['EmployeeNumber'].is_unique, "Validation Error: Duplicate EmployeeNumber found"
assert set(df_attrition['Attrition'].unique()) <= {'Yes', 'No'}, "Validation Error: Invalid Attrition values"

# 2. Performance Data Rules
assert df_perf['Employee ID'].is_unique, "Validation Error: Duplicate Employee ID found"
assert df_perf['Attendance (%)'].between(0, 100).all(), "Validation Error: Attendance out of 0-100 bounds"
assert df_perf['Task Completion (%)'].between(0, 100).all(), "Validation Error: Task Completion out of 0-100 bounds"

# 3. Occupation Master Rules
assert df_occ['O*NET-SOC Code'].is_unique, "Validation Error: Duplicate O*NET-SOC Code found"

print("✅ All data validation checks passed successfully!")
```
