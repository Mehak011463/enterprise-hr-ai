# Data relationships
The supplied performance dataset uses a different employee-ID namespace from the attrition dataset. Do not manufacture a join.

```python
import pandas as pd
p='../data/raw'
a=pd.read_csv(f'{p}/employee_attrition.csv'); perf=pd.read_csv(f'{p}/Employee_Performance_Dataset.csv')
print('Attrition IDs:',a.EmployeeNumber.nunique(),'Performance IDs:',perf['Employee ID'].nunique())
print('Exact ID overlap:',len(set(a.EmployeeNumber)&set(perf['Employee ID'])))
```

The result is zero overlap, so the correct MVP relationship is: attrition data → attrition ML; performance data → independent engagement analytics; O*NET tables → role/skill intelligence.
