```python
import pandas as pd
import os

DATA_PATH = "../data/raw"
OUTPUT_PATH = "../data/processed"
os.makedirs(OUTPUT_PATH, exist_ok=True)

# 1. Standardize Attrition File
df_attrition = pd.read_csv(f"{DATA_PATH}/employee_attrition.csv")
df_attrition = df_attrition.rename(columns={'EmployeeNumber': 'EmployeeID'})
df_attrition.drop_duplicates(inplace=True)
df_attrition.to_csv(f"{OUTPUT_PATH}/employee_attrition_processed.csv", index=False)

# 2. Standardize Performance File
df_perf = pd.read_csv(f"{DATA_PATH}/Employee_Performance_Dataset.csv")
df_perf = df_perf.rename(columns={'Employee ID': 'EmployeeID', 'Job Role': 'JobRole'})
df_perf.drop_duplicates(inplace=True)
df_perf.to_csv(f"{OUTPUT_PATH}/engagement_processed.csv", index=False)

# 3. Standardize Occupation Master
df_occ = pd.read_csv(f"{DATA_PATH}/occupation_data.csv")
df_occ.drop_duplicates(subset=['O*NET-SOC Code'], inplace=True)
df_occ.to_csv(f"{OUTPUT_PATH}/occupation_master.csv", index=False)

# 4. Standardize Essential Skills
df_ess = pd.read_csv(f"{DATA_PATH}/essential_skills.csv")
df_ess.dropna(subset=['O*NET-SOC Code', 'Element Name'], inplace=True)
df_ess.to_csv(f"{OUTPUT_PATH}/essential_skills_processed.csv", index=False)

# 5. Standardize Software Skills
df_soft = pd.read_csv(f"{DATA_PATH}/software_skills.csv")
df_soft.dropna(subset=['O*NET-SOC Code', 'Workplace Example'], inplace=True)
df_soft.to_csv(f"{OUTPUT_PATH}/software_skills_processed.csv", index=False)

print("✅ Data cleaning complete. Cleaned files saved to 'data/processed/'.")
```
