# Recommendation engine
Recommendations are generated from explicit missing skills and a course catalog.

```
import pandas as pd
courses=pd.read_csv('../data/processed/course_catalog.csv')
missing=['MLOps','Docker']
for s in missing:
    print(s, courses[courses.skill.str.lower().eq(s.lower())][['course_name','duration_hours']].to_dict('records'))
```
