```python
import pandas as pd
import numpy as np

processed_path = "../data/processed"

# 1. Load O*NET master datasets
df_occ = pd.read_csv(f"{processed_path}/occupation_master.csv")
df_ess = pd.read_csv(f"{processed_path}/essential_skills_processed.csv")
df_soft = pd.read_csv(f"{processed_path}/software_skills_processed.csv")

print(f"Loaded Occupations: {df_occ.shape[0]}, Essential Skills: {df_ess.shape[0]}, Software Skills: {df_soft.shape[0]}")

# 2. Extract top skills per occupation code
# Filter essential skills where Data Value >= 3.0 (significant requirement)
df_ess_filtered = df_ess[df_ess['Data Value'] >= 3.0]

# Aggregate skill names into a consolidated list per O*NET code
essential_skill_map = df_ess_filtered.groupby('O*NET-SOC Code')['Element Name'].unique().apply(list).to_dict()
software_skill_map = df_soft.groupby('O*NET-SOC Code')['Workplace Example'].unique().apply(list).to_dict()

# 3. Create target role profiles table
df_occ['Required_Essential_Skills'] = df_occ['O*NET-SOC Code'].map(essential_skill_map).fillna("").apply(list)
df_occ['Required_Software_Skills'] = df_occ['O*NET-SOC Code'].map(software_skill_map).fillna("").apply(list)

# 4. Export skill profiles
df_occ.to_csv(f"{processed_path}/role_skill_profiles.csv", index=False)
print("✅ Saved 'role_skill_profiles.csv' successfully!")
```
