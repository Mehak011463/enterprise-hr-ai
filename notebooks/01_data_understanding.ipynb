```python
import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)
DATA_PATH = "../data/raw"

# The 5 primary core datasets from your notes
target_files = {
    "attrition": "employee_attrition.csv",
    "performance": "Employee_Performance_Dataset.csv",
    "occupations": "occupation_data.csv",
    "essential_skills": "essential_skills.csv",
    "software_skills": "software_skills.csv"
}

# Run discovery checks on every file
for key, filename in target_files.items():
    filepath = os.path.join(DATA_PATH, filename)
    print(f"==================== [{key.upper()}] {filename} ====================")
    
    if os.path.exists(filepath):
        df = pd.read_csv(filepath)
        print(f"Shape: {df.shape}")
        print(f"Duplicates: {df.duplicated().sum()}")
        print("\nMissing Values (Top 5):")
        print(df.isnull().sum()[df.isnull().sum() > 0].head(5))
        
        # ID detection
        id_cols = [c for c in df.columns if 'id' in c.lower() or 'code' in c.lower() or 'number' in c.lower()]
        print(f"\nID / Key Candidates: {id_cols}")
        
        # Target distribution check for attrition table
        if key == "attrition" and "Attrition" in df.columns:
            print("\nTarget Class Distribution (%):")
            print(df['Attrition'].value_counts(normalize=True) * 100)
    else:
        print(f"ERROR: File '{filename}' not found in '{DATA_PATH}'.")
    print("\n" + "-"*60 + "\n")
```
