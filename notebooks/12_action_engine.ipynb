```python
import pandas as pd

processed_path = "../data/processed"
df = pd.read_csv(f"{processed_path}/workforce_intelligence_final.csv")

def map_upskilling_pathway(row):
    """
    Maps employee performance and risk profiles to concrete training interventions.
    """
    risk = row['Attrition_Risk_Score']
    sat = row['Overall_Satisfaction']
    
    if risk >= 0.70:
        return "Priority 1: Executive Coaching & Career Pathing Alignment"
    elif risk >= 0.40:
        return "Priority 2: Mid-Level Technical Upskilling & Mentorship"
    elif sat < 2.5:
        return "Priority 3: Workplace Environment & Team Engagement Sync"
    else:
        return "Standard Track: Annual Skill Development Program"

df['Upskilling_Pathway'] = df.apply(map_upskilling_pathway, axis=1)
df.to_csv(f"{processed_path}/workforce_intelligence_final.csv", index=False)
print("✅ Action engine complete. Saved 'Upskilling_Pathway' column.")
```
